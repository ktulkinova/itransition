```python
import time
import sempy.fabric as fabric

model = "air_taxi_model"
```

```python
request_id = fabric.refresh_dataset(model, refresh_type="full")
```

```python
status = "Unknown"
while status == "Unknown":
    time.sleep(10)
    status = fabric.get_refresh_execution_details(model, request_id).status

print(status)
assert status == "Completed"
```
